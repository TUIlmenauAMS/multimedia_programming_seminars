# Chapter 02: WebGL 2

**WebGL 2** is a JavaScript API for rendering interactive 2D and 3D graphics in any modern browser — without plugins. It gives you direct access to the **GPU** (graphics processing unit) through a small, low-level interface based on OpenGL ES 3.0. Everything you see in browser-based games, data visualizations, 3D product viewers and creative coding is ultimately drawn by a WebGL program.

What makes WebGL different from the Web Audio API of the previous chapter is that there is **no ready-made scene graph and no built-in objects**. WebGL does not know what a cube, a camera or a light is. It only knows how to draw **points, lines and triangles** very fast, using small programs called **shaders** that you write yourself and that run in parallel on the GPU. Everything else — transformations, projection, lighting, textures — you build on top of this.

That is exactly why it is worth learning with raw WebGL first: once you understand the pipeline, libraries such as Three.js, Babylon.js or the newer WebGPU API become much easier to understand.

This chapter walks through the core concepts from the first triangle to a lit, textured 3D object and real-time video filters. Each section ends with an interactive demo: a standalone HTML file whose source code is heavily commented, so you can read it like a tutorial.

## 1. What is WebGL?

**OpenGL** is a long-standing, cross-platform graphics API standardized by the Khronos Group. **OpenGL ES** is its slimmed-down version for embedded systems and mobile devices. **WebGL** is a JavaScript binding of OpenGL ES that runs inside the browser and draws onto an HTML `<canvas>` element.

| Version | Based on | Shader language | Status |
| ------- | -------- | --------------- | ------ |
| **WebGL 1** | OpenGL ES 2.0 | GLSL ES 1.00 (`attribute`, `varying`) | Legacy; many older tutorials still use it |
| **WebGL 2** | OpenGL ES 3.0 | GLSL ES 3.00 (`in`, `out`) | Current standard, supported by all major browsers |

### Why WebGL 2 in this course?

WebGL 2 is the modern baseline and is available in current versions of Chrome, Edge, Firefox and Safari. Compared to WebGL 1 it offers:

- **GLSL ES 3.00** with a cleaner syntax (`in` / `out` instead of `attribute` / `varying`)
- **Vertex Array Objects (VAOs)** as a core feature, which make attribute setup much easier to manage
- **Non-power-of-two textures** with mipmaps and repeat wrapping, **3D textures** and **texture arrays**
- **Instanced rendering**, **uniform buffer objects**, **multiple render targets** and **transform feedback** without extensions
- Many more texture and renderbuffer formats

### An outlook: WebGPU

**WebGPU** is the successor of WebGL. It maps to modern GPU APIs (Vulkan, Metal, Direct3D 12), adds **compute shaders** and has a more explicit design. It is still being rolled out across browsers, while WebGL 2 works everywhere today. The concepts you learn in this chapter — buffers, shaders, pipelines, uniforms, textures — carry over directly to WebGPU.

```
WebGL 1  →  WebGL 2  →  WebGPU
(ES 2.0)    (ES 3.0)    (modern, explicit API)
```

## 2. The Rendering Pipeline

The central idea of WebGL is the **rendering pipeline**: a fixed sequence of stages that turns a list of vertices into colored pixels on the screen. Some stages are **programmable** — you write the code — and some are **fixed-function** — you only configure them.

```
Vertex data  →  Vertex Shader  →  Primitive Assembly  →  Rasterization  →  Fragment Shader  →  Per-fragment operations  →  Framebuffer
(your arrays)   (programmable)    (fixed)                 (fixed)           (programmable)       (depth, blending: fixed)    (canvas)
```

| Stage | What happens |
| ----- | ------------ |
| **Vertex shader** | Runs **once per vertex**. Its main job is to compute the vertex position in *clip space* and write it to `gl_Position`. |
| **Primitive assembly** | Groups vertices into primitives (points, lines or triangles) and clips everything outside the visible volume. |
| **Rasterization** | Determines which **pixels** (fragments) are covered by each triangle and interpolates data across it. |
| **Fragment shader** | Runs **once per fragment**. Its job is to compute the final color of that pixel. |
| **Per-fragment operations** | Depth test, blending and other tests decide whether and how the fragment is written to the framebuffer. |

The key property of this design is **massive parallelism**. A GPU has thousands of small cores, and the same vertex shader runs on many vertices at the same time, and the same fragment shader on many pixels at the same time. That is why shaders are small, self-contained programs without shared state.

### Coordinate systems in WebGL

WebGL uses **clip space** for positions: after the vertex shader, x, y and z must be within the range **−1 to +1** (after the perspective divide) to be visible. The center of the canvas is (0, 0), the left edge is x = −1, the right edge x = +1, the bottom edge y = −1 and the top edge y = +1.

```
        y = +1
          │
 x = −1 ──┼── x = +1
          │
        y = −1
```

The **viewport transform** (`gl.viewport`) then maps this −1…+1 range to actual pixel coordinates on the canvas.

## 3. Canvas and WebGL Context

WebGL draws onto a `<canvas>` element. You access the API through a **rendering context** object, which in WebGL 2 is a `WebGL2RenderingContext`. By convention it is stored in a variable called `gl`, and every WebGL function is a method of this object.

```html
<canvas id="glcanvas"></canvas>
```

```javascript
const canvas = document.querySelector('#glcanvas');
const gl = canvas.getContext('webgl2');

if (!gl) {
    // The browser or GPU does not support WebGL 2.
    throw new Error('WebGL 2 is not available.');
}
```

`getContext` returns `null` if WebGL 2 is not supported, so you should always check the result and show a helpful message instead of failing silently.

### Canvas size vs. display size

A canvas has **two sizes**, and mixing them up is the most common beginner mistake:

| Size | Property | Meaning |
| ---- | -------- | ------- |
| **Drawing buffer size** | `canvas.width`, `canvas.height` | The number of pixels WebGL actually renders |
| **Display size** | CSS `width` / `height`, `canvas.clientWidth` | The size of the canvas on the page |

If the two differ, the browser stretches the image and it looks blurry. On high-resolution screens the device pixel ratio (`window.devicePixelRatio`) is often 2 or more, so a professional setup resizes the drawing buffer to match the display:

```javascript
function resizeCanvasToDisplaySize(canvas) {
    const dpr = window.devicePixelRatio || 1;
    const width  = Math.round(canvas.clientWidth  * dpr);
    const height = Math.round(canvas.clientHeight * dpr);

    if (canvas.width !== width || canvas.height !== height) {
        canvas.width  = width;
        canvas.height = height;
    }
}
```

After resizing, tell WebGL how to map clip space to pixels and clear the canvas:

```javascript
gl.viewport(0, 0, gl.canvas.width, gl.canvas.height);
gl.clearColor(0.12, 0.12, 0.18, 1.0);   // r, g, b, a  (0.0 to 1.0)
gl.clear(gl.COLOR_BUFFER_BIT);
```

## 4. Shaders and GLSL

A **shader** is a small program that runs on the GPU. Shaders are written in **GLSL ES 3.00** (OpenGL Shading Language), a C-like language with built-in vector and matrix types such as `vec2`, `vec3`, `vec4` and `mat4`. In WebGL you always need a pair: one **vertex shader** and one **fragment shader**, linked together into a **program**.

### The vertex shader

```glsl
#version 300 es
in vec2 a_position;          // input: one 2D position per vertex

void main() {
    gl_Position = vec4(a_position, 0.0, 1.0);   // required output
}
```

### The fragment shader

```glsl
#version 300 es
precision highp float;       // required in fragment shaders
out vec4 outColor;           // output: the color of this pixel

void main() {
    outColor = vec4(1.0, 0.5, 0.0, 1.0);        // orange, fully opaque
}
```

Note the first line: `#version 300 es` **must** be the very first thing in the source string, with no blank line or whitespace before it.

### Naming conventions

By convention, variables get a prefix that tells you where the data comes from: `a_` for **attributes**, `u_` for **uniforms** and `v_` for **varyings**. This is not required by GLSL, but it makes shader code much easier to read.

### Compiling and linking

The shader source is just a string in JavaScript. It has to be compiled by the GPU driver at runtime:

```javascript
function createShader(gl, type, source) {
    const shader = gl.createShader(type);       // gl.VERTEX_SHADER or gl.FRAGMENT_SHADER
    gl.shaderSource(shader, source);
    gl.compileShader(shader);

    if (!gl.getShaderParameter(shader, gl.COMPILE_STATUS)) {
        const log = gl.getShaderInfoLog(shader);
        gl.deleteShader(shader);
        throw new Error('Shader compile error: ' + log);
    }
    return shader;
}

function createProgram(gl, vertexSource, fragmentSource) {
    const vs = createShader(gl, gl.VERTEX_SHADER, vertexSource);
    const fs = createShader(gl, gl.FRAGMENT_SHADER, fragmentSource);

    const program = gl.createProgram();
    gl.attachShader(program, vs);
    gl.attachShader(program, fs);
    gl.linkProgram(program);

    if (!gl.getProgramParameter(program, gl.LINK_STATUS)) {
        const log = gl.getProgramInfoLog(program);
        gl.deleteProgram(program);
        throw new Error('Program link error: ' + log);
    }
    return program;
}
```

Always check the compile and link status and print the info log. A shader error is otherwise completely silent: you just see a black canvas.

## 5. Buffers, Attributes and VAOs

The geometry data (positions, colors, …) lives in JavaScript arrays and has to be copied to GPU memory. This happens in three steps.

### Step 1 — Buffers

A **buffer** is a block of GPU memory. You create it, bind it and upload typed-array data to it:

```javascript
const positions = new Float32Array([
     0.0,  0.6,     // top vertex
    -0.6, -0.6,     // bottom left
     0.6, -0.6,     // bottom right
]);

const positionBuffer = gl.createBuffer();
gl.bindBuffer(gl.ARRAY_BUFFER, positionBuffer);
gl.bufferData(gl.ARRAY_BUFFER, positions, gl.STATIC_DRAW);
```

WebGL is a **state machine**: `gl.bindBuffer` sets the "current" `ARRAY_BUFFER`, and later calls such as `gl.bufferData` act on whatever is currently bound. `gl.STATIC_DRAW` is a hint that the data is written once and drawn many times.

### Step 2 — Attributes

An **attribute** is a per-vertex input of the vertex shader. You must tell WebGL how to read the buffer: how many numbers per vertex, which type, and where they start.

```javascript
const posLoc = gl.getAttribLocation(program, 'a_position');

gl.enableVertexAttribArray(posLoc);
gl.vertexAttribPointer(
    posLoc,      // which attribute
    2,           // components per vertex (x, y)
    gl.FLOAT,    // data type
    false,       // normalize?
    0,           // stride (0 = tightly packed)
    0            // offset in bytes
);
```

### Step 3 — Vertex Array Objects

In WebGL 2, all of this attribute setup is stored in a **Vertex Array Object (VAO)**. You record the setup once, and at draw time you just bind the VAO again instead of repeating every `vertexAttribPointer` call.

```javascript
const vao = gl.createVertexArray();
gl.bindVertexArray(vao);       // start recording

// ... bindBuffer / enableVertexAttribArray / vertexAttribPointer here ...

gl.bindVertexArray(null);      // stop recording
```

### Drawing

```javascript
gl.useProgram(program);
gl.bindVertexArray(vao);
gl.drawArrays(gl.TRIANGLES, 0, 3);   // primitive type, first vertex, vertex count
```

The primitive type tells WebGL how to group vertices: `gl.POINTS`, `gl.LINES`, `gl.LINE_STRIP`, `gl.TRIANGLES`, `gl.TRIANGLE_STRIP` and `gl.TRIANGLE_FAN`. Triangles are by far the most common, because every surface can be built from them.

**Setup once, draw many times.** Creating shaders, buffers and VAOs is *initialization* and happens once. Only `useProgram`, `bindVertexArray` and the draw call happen every frame.

## 6. Varyings

So far every pixel of the triangle has the same color. To give each vertex its own color you add a second attribute and pass it from the vertex shader to the fragment shader. Such a value is called a **varying** (in GLSL ES 3.00 it is simply an `out` variable in the vertex shader and a matching `in` variable in the fragment shader).

```glsl
#version 300 es
// ---- vertex shader ----
in vec2 a_position;
in vec3 a_color;
out vec3 v_color;            // passed on to the fragment shader

void main() {
    v_color = a_color;
    gl_Position = vec4(a_position, 0.0, 1.0);
}
```

```glsl
#version 300 es
// ---- fragment shader ----
precision highp float;
in vec3 v_color;             // same name and type as the vertex shader's out
out vec4 outColor;

void main() {
    outColor = vec4(v_color, 1.0);
}
```

### Interpolation

The vertex shader runs only three times for a triangle, but the fragment shader runs for **every pixel inside it**. During rasterization the GPU automatically **interpolates** each varying across the triangle using *barycentric coordinates*. A pixel in the middle of an edge between a red and a green vertex receives a 50/50 mix of both colors; a pixel in the center of the triangle receives a mix of all three.

```
      red
      /\
     /  \       →  smooth color gradient
    /    \         across the triangle
 green────blue
```

This interpolation is not limited to colors. The same mechanism is used for **texture coordinates**, **normals** and any other per-vertex data that must vary smoothly across a surface.

For the second attribute you can either use a second buffer or **interleave** the data in one buffer (`x, y, r, g, b, x, y, r, g, b, …`). For interleaved data you set the `stride` to the size of one whole vertex in bytes and the `offset` to where each attribute starts inside it.

```javascript
// Interleaved: 2 floats position + 3 floats color = 5 floats per vertex
const stride = 5 * Float32Array.BYTES_PER_ELEMENT;

gl.vertexAttribPointer(posLoc,   2, gl.FLOAT, false, stride, 0);
gl.vertexAttribPointer(colorLoc, 3, gl.FLOAT, false, stride, 2 * Float32Array.BYTES_PER_ELEMENT);
```

## 7. Uniforms and the Animation Loop

An attribute has a **different value for every vertex**. A **uniform** has the **same value for all vertices and fragments** of one draw call. Uniforms are how JavaScript sends data to a shader: the elapsed time, a color from a slider, the mouse position, a transformation matrix.

```glsl
#version 300 es
precision highp float;
uniform vec3 u_color;        // set from JavaScript
out vec4 outColor;

void main() {
    outColor = vec4(u_color, 1.0);
}
```

You look up the uniform's location once during initialization and then set it with a function that matches its type:

```javascript
const colorLoc = gl.getUniformLocation(program, 'u_color');

gl.useProgram(program);            // the program must be active first
gl.uniform3f(colorLoc, 1.0, 0.5, 0.0);
```

| GLSL type | JavaScript call |
| --------- | --------------- |
| `float` | `gl.uniform1f(loc, x)` |
| `vec2` / `vec3` / `vec4` | `gl.uniform2f` / `gl.uniform3f` / `gl.uniform4f` |
| `int`, `sampler2D` | `gl.uniform1i(loc, n)` |
| `mat3` / `mat4` | `gl.uniformMatrix3fv` / `gl.uniformMatrix4fv` |

### The animation loop

To animate, you redraw the scene about 60 times per second. The browser provides `requestAnimationFrame` for this: it calls your function right before the next screen refresh and passes a high-resolution timestamp in milliseconds.

```javascript
function render(timeMs) {
    const t = timeMs * 0.001;                 // convert to seconds

    resizeCanvasToDisplaySize(gl.canvas);
    gl.viewport(0, 0, gl.canvas.width, gl.canvas.height);
    gl.clear(gl.COLOR_BUFFER_BIT);

    gl.useProgram(program);
    gl.uniform1f(timeLoc, t);                 // new value every frame
    gl.bindVertexArray(vao);
    gl.drawArrays(gl.TRIANGLES, 0, 3);

    requestAnimationFrame(render);            // schedule the next frame
}
requestAnimationFrame(render);
```

Using the **timestamp** instead of a frame counter makes animations run at the same speed on a 60 Hz and a 144 Hz display. Compared with the Web Audio API, the idea is similar: JavaScript is the *control* layer, and the real-time work happens on a separate processor — the audio thread there, the GPU here.

## 8. Transformations and Matrices

To move, rotate or scale an object you do **not** change its vertex data. You keep the geometry as it is and transform it in the vertex shader with a **matrix**. Matrices are the standard tool because they can **combine** several transformations into one, and the GPU multiplies them extremely fast.

### Homogeneous coordinates

A 2D translation cannot be written as a 2×2 matrix multiplication. The trick is to add a third coordinate that is always 1, so a 2D point `(x, y)` becomes `(x, y, 1)`. Then all three basic transformations become 3×3 matrices:

```
Translation              Rotation (angle θ)           Scale
| 1  0  tx |             | cos θ  −sin θ  0 |         | sx  0   0 |
| 0  1  ty |             | sin θ   cos θ  0 |         | 0   sy  0 |
| 0  0  1  |             |   0       0    1 |         | 0   0   1 |
```

The vertex shader multiplies each position by one matrix uniform:

```glsl
#version 300 es
in vec2 a_position;
uniform mat3 u_matrix;

void main() {
    vec3 pos = u_matrix * vec3(a_position, 1.0);
    gl_Position = vec4(pos.xy, 0.0, 1.0);
}
```

### Order matters

Matrices are multiplied from **right to left**. In `M = T · R · S` the vertex is scaled first, then rotated, then translated. Changing the order changes the result: *rotate, then translate* moves an object along a rotated direction, while *translate, then rotate* swings it around the origin.

```
M = Translation × Rotation × Scale
      (last)       (middle)   (first)
```

### Column-major order in JavaScript

WebGL expects matrices as a flat array in **column-major order**: the numbers of the first column come first. The `transpose` argument of `uniformMatrix*` must be `false` in WebGL.

```javascript
function rotation(angle) {
    const c = Math.cos(angle), s = Math.sin(angle);
    return new Float32Array([
         c,  s, 0,     // column 0
        -s,  c, 0,     // column 1
         0,  0, 1,     // column 2
    ]);
}

gl.uniformMatrix3fv(matrixLoc, false, rotation(t));
```

In practice, most projects use a library such as **gl-matrix** for matrix math. In our demos we write a small helper ourselves, so you can see exactly what happens.

## 9. Going 3D

WebGL itself is always a 2D rasterizer. 3D is an **illusion** created by the vertex shader, which projects 3D positions onto the 2D screen. This is done with three matrices that are combined into one.

### Model, View and Projection

| Matrix | Purpose |
| ------ | ------- |
| **Model** | Places the object in the world: its position, rotation and scale |
| **View** | Moves the world so that the camera sits at the origin and looks along −z |
| **Projection** | Turns the camera's view volume into clip space, creating perspective |

```glsl
#version 300 es
in vec3 a_position;
uniform mat4 u_model;
uniform mat4 u_view;
uniform mat4 u_projection;

void main() {
    gl_Position = u_projection * u_view * u_model * vec4(a_position, 1.0);
}
```

The **perspective projection** is defined by the field of view, the aspect ratio of the canvas and the near and far clipping planes. After the vertex shader, the GPU automatically divides x, y and z by the fourth component `w` (the **perspective divide**). Objects far away have a large `w`, so they end up smaller on screen.

```javascript
function perspective(fovY, aspect, near, far) {
    const f  = 1 / Math.tan(fovY / 2);
    const nf = 1 / (near - far);
    return new Float32Array([
        f / aspect, 0, 0,                    0,
        0,          f, 0,                    0,
        0,          0, (far + near) * nf,   -1,
        0,          0, 2 * far * near * nf,  0,
    ]);
}
```

### The depth test

When several triangles cover the same pixel, WebGL must decide which one is in front. Without any help it simply draws them in the order you submit them, so a back face can paint over a front face. The **depth buffer** solves this: it stores the depth of the closest fragment drawn so far for every pixel, and discards fragments that are farther away.

```javascript
gl.enable(gl.DEPTH_TEST);

// Clear both color and depth at the start of every frame
gl.clear(gl.COLOR_BUFFER_BIT | gl.DEPTH_BUFFER_BIT);
```

You can also enable **back-face culling** with `gl.enable(gl.CULL_FACE)`. Triangles whose vertices appear in clockwise order on screen are then skipped. By default, **counter-clockwise** winding means front-facing.

### Index buffers

A cube has 8 corners but 12 triangles, which means 36 vertices if every triangle is listed separately. An **index buffer** (`ELEMENT_ARRAY_BUFFER`) lists each vertex once and then refers to vertices by number, so shared corners are not duplicated.

```javascript
const indices = new Uint16Array([
    0, 1, 2,   0, 2, 3,     // first face: two triangles
    // ... more faces ...
]);

const indexBuffer = gl.createBuffer();
gl.bindBuffer(gl.ELEMENT_ARRAY_BUFFER, indexBuffer);
gl.bufferData(gl.ELEMENT_ARRAY_BUFFER, indices, gl.STATIC_DRAW);

// At draw time:
gl.drawElements(gl.TRIANGLES, indices.length, gl.UNSIGNED_SHORT, 0);
```

The index buffer binding is stored in the **VAO**, so bind the VAO first and then create the index buffer. In practice, if each face needs its own color, normal or texture coordinate, a cube uses **24 vertices** (4 per face) so that a corner can have different data on each face it belongs to.

## 10. Textures

A **texture** is an image stored in GPU memory that shaders can read from. Instead of giving each vertex a color, you give it a **texture coordinate** (also called UV coordinate) that says which point of the image belongs to that vertex. The GPU interpolates these coordinates across the triangle, just like a varying color.

```
(0,1) ┌───────────┐ (1,1)
      │           │            u goes left → right
      │  texture  │            v goes bottom → top
      │           │
(0,0) └───────────┘ (1,0)
```

Texture coordinates always range from 0 to 1, independent of the image size in pixels.

### Creating a texture

```javascript
const texture = gl.createTexture();
gl.bindTexture(gl.TEXTURE_2D, texture);

// Upload the pixels. The source can be an <img>, a <canvas>, a <video> or typed array data.
gl.texImage2D(gl.TEXTURE_2D, 0, gl.RGBA, gl.RGBA, gl.UNSIGNED_BYTE, sourceCanvas);

gl.generateMipmap(gl.TEXTURE_2D);
```

HTML images have their origin at the **top left**, while texture coordinates start at the **bottom left**. To compensate, call `gl.pixelStorei(gl.UNPACK_FLIP_Y_WEBGL, true)` before uploading.

### Filtering and wrapping

| Parameter | Options | Effect |
| --------- | ------- | ------ |
| `TEXTURE_MAG_FILTER` | `NEAREST`, `LINEAR` | Used when the texture is magnified: blocky pixels or smooth blending |
| `TEXTURE_MIN_FILTER` | `NEAREST`, `LINEAR`, `LINEAR_MIPMAP_LINEAR`, … | Used when the texture is shrunk. Mipmaps (pre-scaled smaller copies) avoid flickering and aliasing |
| `TEXTURE_WRAP_S/T` | `CLAMP_TO_EDGE`, `REPEAT`, `MIRRORED_REPEAT` | What happens for coordinates outside 0…1 |

```javascript
gl.texParameteri(gl.TEXTURE_2D, gl.TEXTURE_MIN_FILTER, gl.LINEAR_MIPMAP_LINEAR);
gl.texParameteri(gl.TEXTURE_2D, gl.TEXTURE_MAG_FILTER, gl.LINEAR);
gl.texParameteri(gl.TEXTURE_2D, gl.TEXTURE_WRAP_S, gl.REPEAT);
gl.texParameteri(gl.TEXTURE_2D, gl.TEXTURE_WRAP_T, gl.REPEAT);
```

In WebGL 1 mipmaps and `REPEAT` only worked for power-of-two image sizes. In WebGL 2 this restriction is gone.

### Sampling in the shader

A texture is read in the fragment shader through a **sampler** uniform. The texture is bound to a **texture unit**, and the sampler uniform holds the number of that unit.

```glsl
#version 300 es
precision highp float;
in vec2 v_uv;                  // interpolated texture coordinate
uniform sampler2D u_texture;
out vec4 outColor;

void main() {
    outColor = texture(u_texture, v_uv);
}
```

```javascript
gl.activeTexture(gl.TEXTURE0);              // select texture unit 0
gl.bindTexture(gl.TEXTURE_2D, texture);
gl.uniform1i(textureLoc, 0);                // sampler reads from unit 0
```

In our texture demo the image is generated in JavaScript on a 2D canvas, so there is no external file and no loading or cross-origin issues when you open the file locally.

### Video as a texture

`texImage2D` also accepts an `HTMLVideoElement`. A video is simply **a texture whose pixels change every frame**: you keep the `<video>` element playing in the background and copy its current frame to the texture once per frame in the render loop.

```javascript
function render() {
    // Upload only when the video has a frame available
    if (video.readyState >= video.HAVE_CURRENT_DATA) {
        gl.bindTexture(gl.TEXTURE_2D, texture);
        gl.texImage2D(gl.TEXTURE_2D, 0, gl.RGBA, gl.RGBA, gl.UNSIGNED_BYTE, video);
    }

    gl.drawArrays(gl.TRIANGLE_STRIP, 0, 4);
    requestAnimationFrame(render);
}
```

For video, the texture is usually set up with `LINEAR` filtering, `CLAMP_TO_EDGE` wrapping and **no mipmaps**: mipmaps would have to be regenerated every frame, and a video is typically shown at about its original size.

**Video sources.** A local file is played through `URL.createObjectURL(file)`; the webcam is available through `navigator.mediaDevices.getUserMedia({ video: true })` and needs the user's permission. For autoplay the video element must be `muted`.

### Real-time filters in the fragment shader

Because the fragment shader runs once per pixel, in parallel, it is ideal for image and video processing. A 1080p frame has about 2 million pixels, and the GPU processes all of them at the same time. The vertex shader only draws a quad; all the work happens in the fragment shader.

```glsl
uniform sampler2D u_texture;

vec3 grayscale(vec3 c) {
    // Luminance weights for Rec. 709: the eye is most sensitive to green
    float y = dot(c, vec3(0.2126, 0.7152, 0.0722));
    return vec3(y);
}

void main() {
    vec3 color = texture(u_texture, v_uv).rgb;
    outColor = vec4(grayscale(color), 1.0);
}
```

Filters that need neighbouring pixels, such as **blur** or **Sobel edge detection**, sample the texture several times with small offsets. In GLSL ES 3.00 the size of one texel is `1.0 / vec2(textureSize(u_texture, 0))`.

## 11. Lighting

WebGL has **no built-in lighting**. The old fixed-function OpenGL had it, modern APIs do not. Lighting is simply math in your shaders. This section is an outlook for the optional demo.

### Normals

To compute how much light hits a surface, the shader needs the surface **normal**: a unit vector pointing perpendicular to the surface. Normals are provided as an additional attribute, one per vertex.

### Diffuse (Lambert) lighting

A matte surface is brightest when it faces the light and darkens as it turns away. This is described by the **dot product** of the normal **N** and the direction to the light **L**, both normalized:

```
diffuse = max( dot(N, L), 0 )
```

If N and L point in the same direction the dot product is 1 (full brightness). At 90° it is 0, and for surfaces facing away it is negative, so we clamp it to 0.

```glsl
#version 300 es
precision highp float;
in vec3 v_normal;
uniform vec3 u_lightDirection;   // direction pointing TOWARD the light
uniform vec3 u_baseColor;
uniform float u_ambient;         // minimum brightness, e.g. 0.15
out vec4 outColor;

void main() {
    vec3 N = normalize(v_normal);    // interpolation can shorten the vector
    vec3 L = normalize(u_lightDirection);

    float diffuse = max(dot(N, L), 0.0);
    vec3 color = u_baseColor * (u_ambient + diffuse);

    outColor = vec4(color, 1.0);
}
```

### The normal matrix

When an object is rotated, its normals must rotate as well. For rotations the model matrix works fine. For non-uniform scaling, normals must be transformed with the **inverse transpose** of the upper-left 3×3 part of the model matrix (the *normal matrix*), otherwise they stop being perpendicular to the surface.

### Beyond diffuse

Real-time shading models build on this foundation: **ambient** (constant base light), **diffuse** (this section), **specular** (highlights, depending on the viewer's position — the *Phong* and *Blinn-Phong* models) and, in modern engines, **physically based rendering (PBR)**.
