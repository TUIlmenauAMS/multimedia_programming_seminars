# Welcome to Multimediaprogrammierung

Welcome to the first class of **Multimediaprogrammierung** (Multimedia Coding). In this course you will learn how digital multimedia, such as audio, images and video, is represented, processed and coded, and you will turn that knowledge into working software.

This introductory chapter presents the tools, platforms and resources we will use throughout the semester, and the first task you will complete.

### Goal of the course

By the end of the course you will be able to write software, either a **desktop application**, a **mobile application** or a **web application**, using the knowledge acquired in class.

### Evaluation

You will be evaluated on two things:

1. **Your software**, the application you develop during the course.
2. **A scientific paper** describing your software, written with the official **IEEE template**.

## 2. IEEE

The **IEEE** (Institute of Electrical and Electronics Engineers) is the world's largest technical professional organization. It publishes a large share of the world's research in electrical engineering, computer science and multimedia technology, and it defines many technical standards. In this course it matters in two ways: it is our main source of **scientific literature**, and its **paper template** is the format in which you will write your final paper.

| **IEEE Xplore**       | The digital library of the IEEE, giving access to millions of journal articles, conference papers and standards. This is where you will search for the paper in your first task. Access is available through the university network. |
| **IEEE Audio Coding** | A topic page collecting publications on audio coding: the compression of audio signals, from perceptual codecs such as MP3 and AAC to modern neural audio codecs.                                                                    |

## 5. Linux (Ubuntu)

It runs most of the world's servers and a large share of the tools used in software development, and many multimedia and animation programs are developed for it first. **Ubuntu** is one of the most popular Linux distributions. It is free, well documented and has a large community, and its long-term support (LTS) releases receive updates for years. This makes it a common choice for people who are new to Linux.

## 6. Windows Subsystem for Linux (WSL)

The **Windows Subsystem for Linux (WSL)** is a feature of Windows that lets you run a real Linux environment, such as Ubuntu, directly on a Windows computer. You get the Linux terminal and Linux development tools without leaving Windows, and without setting up a separate virtual machine or a dual-boot installation. Microsoft's documentation explains what WSL is and how to install it.

## 7. Git, GitHub and GitLab

**Git** is a version control system. It records every change to your files, lets you go back to any earlier state, and allows many people to work on the same project without overwriting each other's work. It is the standard tool in software development, and you will use it for your project.

**GitHub** and **GitLab** are web platforms that host Git repositories. They add collaboration features such as issue tracking, code review and project pages, and they make it easy to back up and share your code. The two are very similar in how they work.

- The AMS group publishes its teaching material and code on **GitHub**.
- The university runs its own **GitLab** server, available to every student with a university account.

## 9. Jupyter Notebooks and Google Colab

A **Jupyter Notebook** is an interactive document that combines explanatory text, executable code and results such as plots, images and audio in one place. This is the format of the document you are reading right now. It is ideal for learning, experimenting and presenting results.

**Google Colab** (Colaboratory) runs Jupyter Notebooks in the cloud, for free, directly in your browser. There is nothing to install, and your notebooks are saved to your Google Drive. It is the quickest way to start, and it is a useful fallback when something does not work on your own computer.

## 10. IDE and Visual Studio Code

An **IDE** (Integrated Development Environment) is a program that brings together everything you need to write software: a code editor with syntax highlighting and auto-completion, a debugger, a terminal and tools for version control. It makes you faster and helps you find errors early.

**Visual Studio Code** (VS Code) is a free, lightweight and very popular editor with extensions for almost every language. With the Python and Jupyter extensions it can run Python scripts and notebooks, work with Git and manage virtual environments. It can also connect to a WSL environment. We recommend VS Code for the course, and it is also the way to develop larger projects such as your final application.

## 11. Virtual Environment and uv

A **virtual environment** is an isolated folder holding a project's own copy of Python and its libraries. Different projects often need different library versions, and installing everything globally leads to conflicts. With one environment per project, your setup stays clean, and anyone can reproduce it on their own machine. This matters for your final software, because others should be able to run it.

**uv** is a fast, modern tool for Python that creates virtual environments, installs packages and manages Python versions and project dependencies, all in one program. It replaces the combination of `pip`, `venv` and several other tools with a single simple workflow, and it is much faster than the traditional tools.

## 12. Claude

**Claude** is an AI assistant developed by Anthropic. You can have a conversation with it in plain language, and it can read, write and explain text and code. For this course it is useful in two areas.

- **Research:** Claude can explain difficult concepts in a paper, summarize a topic and help you find the right questions to ask. Always verify its statements against the original sources.
- **Coding:** Claude can help you understand error messages, explain what a piece of code does, suggest approaches and review your code. It works best when you describe your problem clearly.

Claude is a helper for learning, not a replacement for it. You are responsible for your own work, so make sure you understand every piece of code that ends up in your project.
